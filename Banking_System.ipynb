```python
import random

bank_account = {}

def account_number():
    while True:
        acc = random.randint(1000, 9999)
        if acc not in bank_account:
            return acc

def get_float(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print(" Please enter a valid number.")

def get_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print(" Please enter a valid integer.")

def new_user():
    try:
        name = input("Enter your name: ").strip()
        deposit = get_float("Enter initial deposit: ")

        while True:
            pin = input("Set your 4-digit PIN: ")
            if pin.isdigit() and len(pin) == 4:
                pin = int(pin)
                break
            else:
                print(" PIN must be exactly 4 digits.")

        acc_no = account_number()

        bank_account[acc_no] = {
            "name": name,
            "balance": deposit,
            "pin": pin
        }

        print(" Account created successfully!")
        print(f"Your Account Number is: {acc_no}")

    except Exception as e:
        print(" Something went wrong:", e)

    main_menu()


def existing_user():
    try:
        acc_no = get_int("Enter your Account Number: ")
        pin = get_int("Enter your PIN: ")

        if acc_no in bank_account and bank_account[acc_no]["pin"] == pin:
            print(f" Welcome {bank_account[acc_no]['name']}")
            user_menu(acc_no)
        else:
            print(" Invalid Account Number or PIN")

    except Exception as e:
        print(" Error:", e)

    main_menu()


def user_menu(acc_no):
    while True:
        print("\n1. Deposit")
        print("2. Withdraw")
        print("3. Check Balance")
        print("4. Exit")

        choice = input("Enter choice: ")

        if choice == '1':
            amount = get_float("Enter amount: ")
            if amount > 0:
                bank_account[acc_no]["balance"] += amount
                print(" Deposit successful. Balance:", bank_account[acc_no]["balance"])
            else:
                print(" Amount must be positive")

        elif choice == '2':
            amount = get_float("Enter amount: ")
            if amount <= 0:
                print(" Amount must be positive")
            elif amount <= bank_account[acc_no]["balance"]:
                bank_account[acc_no]["balance"] -= amount
                print(" Withdrawal successful. Balance:", bank_account[acc_no]["balance"])
            else:
                print(" Insufficient balance")

        elif choice == '3':
            print(f" Balance: {bank_account[acc_no]['balance']}")

        elif choice == '4':
            print(" Logging out...")
            break

        else:
            print(" Invalid choice")

    main_menu()


def main_menu():
    while True:
        print("\n===== Welcome to the Bank =====")
        print("1. New User")
        print("2. Existing User")
        print("3. Exit")

        choice = input("Enter choice: ")

        if choice == '1':
            new_user()
            break

        elif choice == '2':
            existing_user()
            break

        elif choice == '3':
            print(" Thank you for using our services!")
            break

        else:
            print(" Invalid choice. Try again.")


main_menu()
```
